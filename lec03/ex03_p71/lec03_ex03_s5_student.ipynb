## 1. 문제상황

- 구름월드 고객센터 검색기는 임계값 1.5를 쓰고 있습니다. 이 숫자는 다른 문서에서 가져온 값이라, 구름월드 FAQ에 맞는지 아무도 확인하지 않았습니다.
- 시험 운영에서 문서 안 질문인데 컷되는 일과, 문서 밖 질문인데 통과되는 일이 드물게 생겼습니다.
- 담당자는 문서 안 질문 여러 개와 문서 밖 질문 여러 개의 1위 점수를 실제로 재서, 두 그룹 사이 어디에 선을 그어야 하는지 눈으로 보고 임계값을 정하고 싶어 합니다.

## 2. 문제와 목표

- **문제**: 임계값 1.5가 이 문서의 점수 분포에 맞는지 확인되지 않았습니다.
- **목표**: 구름월드 FAQ 검색기를 세운 뒤, 문서 안 질문 5개와 문서 밖 질문 5개의 1위 점수를 재서 표로 출력하고, 문서 안 점수의 최댓값과 문서 밖 점수의 최솟값 사이의 중간을 새 임계값으로 정해, 새 임계값으로 10개 질문을 다시 판정하는 프로그램을 만듭니다.
    - FAQ 파일: `day05_faq_구름월드.csv`(32행). 저장 디렉터리: `chroma_db`.
    - 문서 안 질문 5개와 문서 밖 질문 5개: 「6. 코드 — 스텝바이스텝」 단계 0에 주어져 있습니다.
    - 고정 안내 문장: 「문서에서 근거를 찾지 못했습니다. 안내 창구로 문의해 주세요.」
    - 출력 줄: 「겹침 없음: True/False」와 「문서 안 통과 N/5 · 문서 밖 컷 N/5」.
- **목표 달성 여부의 판정 기준**: 문서 안 점수의 최댓값이 문서 밖 점수의 최솟값보다 작고(두 그룹이 겹치지 않음), 새 임계값이 그 사이에 있으며, 새 임계값으로 판정했을 때 문서 안 5개가 모두 통과·문서 밖 5개가 모두 컷되는 것을 출력에서 확인합니다.
