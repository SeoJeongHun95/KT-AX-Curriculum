## 1. 문제상황

- 구름월드 상담 화면은 고객이 질문을 보내면 답이 다 만들어질 때까지 빈 화면을 보여 줍니다.
- 「환불 규정을 처음부터 끝까지 자세히 설명해 주세요」처럼 답이 긴 질문에서는 빈 화면이 여러 초 이어지고, 고객은 프로그램이 멈춘 줄 알고 질문을 다시 보냅니다.
- 답이 만들어지는 동안 조금씩이라도 화면에 글자가 나타나면 고객은 기다립니다.
- 담당자는 같은 체인을 그대로 두고, 답을 조각으로 받아 첫 조각이 몇 초 만에 도착하는지, 전체를 한 번에 받을 때와 얼마나 다른지 재 보기를 원합니다.

## 2. 문제와 목표

- **문제**: 답 전체가 완성될 때까지 화면이 비어 있어 고객이 기다리지 못합니다.
- **목표**
  - 같은 부품으로 텍스트 체인을 만듭니다.
  - 전체를 한 번에 받는 실행과 조각으로 받는 실행의 소요 시간을 재서 비교하는 프로그램을 만듭니다.
    - 재는 것 셋: 전체 시간, 첫 조각 도착 시간, 조각 수
  - 주어지는 것: FAQ 문자열 세 항목(운영시간·주차·환불)과 시험 질문 「환불 규정을 처음부터 끝까지 자세히 설명해 주세요.」. FAQ 문자열의 값은 「6. 코드 — 스텝바이스텝」 단계 0에 주어져 있습니다.
  - 모델·템플릿·스키마는 시연과 같은 구성(주제·답변 두 칸 스키마, 시설 안내 담당자 시스템 메시지)이며, 조각 수신에는 모델을 `streaming=True`로 만듭니다.
  - 시간은 `time.perf_counter()`를 호출 직전과 직후에 받아 뺍니다. 첫 조각 도착 시간은 첫 조각을 받은 직후에 받습니다.
- **목표 달성 여부의 판정 기준**:
  - 긴 답이 나오는 질문에서 조각 실행의 첫 조각 도착 시간이 한 번에 받는 실행의 전체 소요 시간보다 짧고,
  - 조각 수가 2개 이상이며, 두 방식이 돌려준 답의 글자 수가 비슷한 것을 출력에서 확인합니다.
  - 출력에는 한 번에 받는 실행의 전체 시간, 조각 실행의 첫 조각 도착 시간과 조각 수, 두 답의 글자 수가 찍힙니다.
