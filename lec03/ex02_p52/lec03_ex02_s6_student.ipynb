## 1. 문제상황

- 구름월드 고객센터는 밤사이 쌓인 미답변 질문을 아침에 한꺼번에 처리합니다. 보통 스무 건 안팎입니다.
- 지금의 구조화 출력 체인으로 질문을 하나씩 차례로 돌리면, 질문 한 건에 몇 초씩 걸려 스무 건이면 1분 넘게 기다립니다.
- 질문들은 서로 관계가 없어 순서대로 처리할 이유가 없는데도, 앞 질문의 답이 올 때까지 뒤 질문은 시작조차 못 합니다.
- 담당자는 질문 여러 건을 한꺼번에 걸어 두고, 답이 오는 대로 모아 표로 정리하고, 하나씩 돌릴 때보다 얼마나 빨라졌는지 알고 싶어 합니다.

## 2. 문제와 목표

- **문제**: 서로 관계없는 질문들을 하나씩 차례로 돌려 소요 시간이 질문 수만큼 더해집니다.
- **목표**
  - 주제·답변 두 칸의 구조화 출력 체인을 그대로 두고, 질문 여러 건을 비동기 실행으로 동시에 걸어 파이썬 표준 비동기 라이브러리로 한꺼번에 기다립니다.
    - 두 실행: 차례 실행(질문을 하나씩), 동시 실행(질문 5건을 한꺼번에 걸고 한 번에 기다림)
  - 주제·답변 표로 정리하고 하나씩 돌릴 때와 소요 시간을 비교하는 프로그램을 만듭니다.
  - 주어지는 것: FAQ 문자열 세 항목과 질문 5건. 값은 「6. 코드 — 스텝바이스텝」 단계 0에 주어져 있습니다.
  - 체인은 시연과 같은 구조화 출력 체인(주제·답변 두 칸)입니다.
  - 동시 실행은 `chain.ainvoke(...)` 다섯 개를 `asyncio.gather(...)`로 한꺼번에 기다립니다. 사용법은 검색해 익힙니다.
- **목표 달성 여부의 판정 기준**:
  - 질문 5건에서 동시 실행의 소요 시간이 차례 실행의 절반 이하이고,
  - 5건 모두 구조화 객체로 돌아와 질문 순서대로 표에 찍히는 것을 출력에서 확인합니다.
  - 출력에는 차례 실행과 동시 실행의 소요 시간, 질문 순서대로 주제·답변을 적은 표가 찍힙니다.
