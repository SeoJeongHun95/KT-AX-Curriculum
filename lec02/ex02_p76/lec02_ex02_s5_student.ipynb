## 1. 문제상황

- 새 프로젝트의 킥오프 회의 메모를 결정사항·할 일·미결로 요약하는 요약기에 넣었더니, 할 일의 기한이 「다음 주 화요일까지」「이번 달 말까지」처럼 메모에 적힌 말 그대로 돌아왔습니다.
- 회의록 담당자는 그 기한을 달력에 옮겨 적어야 하는데, 「화요일」이 어느 화요일인지는 회의 날짜를 아는 사람만 압니다.
- 기한을 날짜순으로 정렬하려 해도 문자열이라 정렬할 수 없고, 달력 프로그램에 넣으면 형식 오류가 납니다.
- 모양 검사는 통과합니다. 기한 필드에 문자열이 들어 있기 때문입니다. 값 검증기를 달면 걸리지만, 걸린 뒤에 어떻게 할지가 없습니다.
- 검증 실패를 사람이 받아 모델에게 다시 호출하는 일이 회의마다 반복됩니다.

## 2. 문제와 목표

- **문제**: 기한이 상대 표현의 문자열로 돌아오고, 값 검증기가 걸러 내더라도 실패한 뒤의 처리가 없습니다. 검증 실패가 프로그램의 끝이 됩니다.
- **목표**: 기한이 달력 날짜(`YYYY-MM-DD`)이거나 없음이어야만 통과하도록 검증을 엄격히 하고, 검증에 실패하면 실패 이유와 회의 날짜를 모델에게 알려 다시 호출하는 처리를 프로그램 안에 넣습니다. 다시 호출하는 횟수에는 상한을 둡니다. 통과한 요약에서 할 일을 기한순으로 정렬해 출력합니다.
    - 회의 날짜: 2026-09-04(금요일). 회의 메모 원문은 「6. 코드 — 스텝바이스텝」 단계 0에 주어져 있습니다.
    - 다시 호출하는 횟수의 상한: 2회.
    - 통과하는 기한: `YYYY-MM-DD` 형식의 달력 날짜 또는 없음(None).
- **목표 달성 여부의 판정 기준**: 회의 메모를 넣었을 때 첫 답이 검증에 걸리고, 다시 받은 답이 통과해 요약 객체가 되며, 할 일이 달력 날짜순으로 출력되는 것을 실행 기록에서 확인합니다. 상한 안에 통과하지 못하면 프로그램이 명시적으로 멈춥니다.
